## How to setup and use MLFlow
We use github for code and MLFlow (on dagshub) for tracking hyperparameters and results

### Do this once:

- make sure you acceptd the invitation sent
- pip install requirements under this folder
- pip install pytorch (run this in cmd line: pip3 install torch torchvision)


### Always do this before an experiment:

```python
import dagshub
import mlflow

dagshub.init(
    repo_owner="HindAlz",
    repo_name="Sequential-Diagnosis",
    mlflow=True,
)

mlflow.set_experiment("sdp-sequential-diagnosis")
```
